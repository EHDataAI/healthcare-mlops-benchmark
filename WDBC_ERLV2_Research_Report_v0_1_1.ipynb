<div class="erl-section">02 · Diseño metodológico</div>
## Particiones estrictas y calibración sin reusar validaciones

El dataset público **Wisconsin Diagnostic Breast Cancer** contiene 569 observaciones y 30 características cuantitativas extraídas de imágenes de aspiración con aguja fina (FNA). La clase positiva corresponde a **malignidad (`1`)**. El experimento trabaja únicamente con 455 ejemplos de desarrollo: 170 malignos y 285 benignos.

**Protocolo ERL-03C v0.3.1 aprobado**

1. **CV externa:** cinco particiones estratificadas de 364 muestras de entrenamiento y 91 de validación; los resultados de validación externa a cada fold no intervienen en tuning, calibración ni umbral.
2. **Búsqueda de parámetros:** tres folds internos; las variantes se comparan mediante average precision (AP).
3. **OOF de umbral:** tres subfolds; en cada subfold, los hiperparámetros se vuelven a optimizar **dentro de su propio entrenamiento**, nunca en sus 121/122 observaciones reservadas para el umbral.
4. **Calibración:** `CalibratedClassifierCV(method="sigmoid", ensemble=False)` con tres folds internos exclusivos del entrenamiento.
5. **Predicción evaluable:** cada caso recibe exactamente una predicción OOF externa por procedimiento; no se ajusta el calibrador sobre los OOF externos que se evalúan.

<div class="erl-note"><strong>Sobre el 95 %:</strong> la sensibilidad ≥95 % es una restricción ilustrativa para seleccionar umbrales *internos*, no una garantía de sensibilidad en validación externa por fold ni en práctica clínica.</div>

| Fase | Datos utilizados | Salida fijada |
|:--|:--|:--|
| ERL-02D | Desarrollo; 5×3 folds | Evaluación exploratoria de 7 familias |
| ERL-03B | Desarrollo; mismas particiones | 3.185 predicciones RAW auditadas |
| ERL-03C | Desarrollo; tuning anidado y calibración interna | 1.820 predicciones (LR/CatBoost × RAW/SIGMOID) |
| Holdout histórico | **Sin acceso en este notebook** | No apto como nueva validación |

```python
LABELS = {
    "dummy_prior": "Referencia por prevalencia",
    "logistic_regression": "Regresión logística",
    "logistic_elastic_net": "Logística ElasticNet",
    "svm_rbf": "SVM · RBF",
    "random_forest": "Random Forest",
    "xgboost": "XGBoost",
    "catboost": "CatBoost",
}
FAMILIES_03C = ("logistic_regression", "catboost")
METHODS = ("raw", "sigmoid")


def group_rows(rows: list[dict]) -> dict[tuple[str, str], list[dict]]:
    groups = {}
    for row in rows:
        groups.setdefault((row["model_family"], row["calibration_method"]), []).append(row)
    return groups


def check_partition(rows: list[dict], n_per_group: int) -> dict:
    groups = group_rows(rows)
    for key, obs in groups.items():
        ids = [r["development_row_id"] for r in obs]
        if len(obs) != n_per_group or set(ids) != set(range(n_per_group)):
            raise ValueError(f"Cobertura OOF incorrecta: {key}")
        if len(ids) != len(set(ids)):
            raise ValueError(f"Identificador OOF repetido: {key}")
        y = [r["y_true_malignant"] for r in obs]
        if Counter(y) != {0: 285, 1: 170}:
            raise ValueError(f"Clases inconsistentes: {key}")
        folds = Counter(r["fold_id"] for r in obs)
        if folds != {i: 91 for i in range(5)}:
            raise ValueError(f"Folds inconsistentes: {key}")
        for row in obs:
            p, cutoff = row["p_malignant"], row["threshold_train_only"]
            if not (math.isfinite(p) and math.isfinite(cutoff)
                    and 0 <= p <= 1 and 0 <= cutoff <= 1):
                raise ValueError(f"Score o umbral inválido: {key}")
            if row["predicted_malignant"] != int(p >= cutoff):
                raise ValueError(f"Decisión inconsistente: {key}")
    return groups


def metrics(obs: list[dict]) -> dict[str, float | int]:
    y = np.array([r["y_true_malignant"] for r in obs], dtype=int)
    p = np.array([r["p_malignant"] for r in obs], dtype=float)
    pred = np.array([r["predicted_malignant"] for r in obs], dtype=int)
    tp = int(((pred == 1) & (y == 1)).sum())
    fn = int(((pred == 0) & (y == 1)).sum())
    tn = int(((pred == 0) & (y == 0)).sum())
    fp = int(((pred == 1) & (y == 0)).sum())
    pclip = np.clip(p, 1e-15, 1 - 1e-15)
    return {
        "sensitivity_malignant": tp / (tp + fn),
        "specificity_benign": tn / (tn + fp),
        "false_negatives_malignant": fn,
        "false_positives_benign": fp,
        "average_precision": float(average_precision_score(y, p)),
        "roc_auc": float(roc_auc_score(y, p)),
        "brier": float(np.mean((p-y) ** 2)),
        "log_loss": float(np.mean(-(y * np.log(pclip) + (1-y) * np.log(1-pclip)))),
    }


def verify_metrics(actual: dict, recorded: dict, label: str) -> None:
    for name, observed in actual.items():
        if not math.isclose(observed, recorded[name], rel_tol=1e-10, abs_tol=1e-12):
            raise ValueError(f"Métrica no reproducible {label} / {name}")


pairs_03b = check_partition(oof_03b, 455)
pairs_03c = check_partition(oof_03c, 455)
assert set(pairs_03b) == {(k, "raw") for k in LABELS}
assert set(pairs_03c) == {(k, m) for k in FAMILIES_03C for m in METHODS}
assert len(oof_03b) == 3185 and len(oof_03c) == 1820

# Consistencia de etiquetas, folds e identificadores para todas las familias/métodos.
reference = {r["development_row_id"]: (r["y_true_malignant"], r["fold_id"])
             for r in pairs_03b[("dummy_prior", "raw")]}
for observation in oof_03b + oof_03c:
    k = observation["development_row_id"]
    assert reference[k] == (observation["y_true_malignant"], observation["fold_id"])

for (family, method), obs in pairs_03c.items():
    verify_metrics(metrics(obs), report_03c["comparison"][f"{family}/{method}"]["metrics"],
                   f"{family}/{method}")
    for fold in range(5):
        part = [r for r in obs if r["fold_id"] == fold]
        assert Counter(r["y_true_malignant"] for r in part) == {0: 57, 1: 34}
        recorded = report_03c["fold_details"][fold]["methods"][f"{family}/{method}"]["metrics"]
        verify_metrics(metrics(part), recorded, f"{family}/{method}/fold{fold}")

for (family, _method), obs in pairs_03b.items():
    recorded = report_02["candidate_results"][family]["nested_oof_metrics"]
    verify_metrics(metrics(obs), recorded, f"ERL03B/{family}")

for family in FAMILIES_03C:
    prev = {r["development_row_id"]: r for r in pairs_03b[(family, "raw")]}
    for row in pairs_03c[(family, "raw")]:
        other = prev[row["development_row_id"]]
        assert row["p_malignant"] == other["p_malignant"]

print("ERL GATE 03 · Cobertura, clases, 5 folds, métricas y paridad RAW PASS")
```

<div class="erl-section">03 · Benchmark exploratorio</div>
## Siete familias. Un mismo marco experimental.

Esta comparación conserva el alcance y las limitaciones originales de ERL-02D/03B. La elección exploratoria fue regresión logística; **no es una selección validada externamente**. Los scores y errores se reconstruyen desde las 3.185 predicciones RAW, mientras que el reporte de benchmark sirve para contrastar las métricas.

```python
BENCHMARK_ORDER = (
    "logistic_regression", "logistic_elastic_net", "svm_rbf", "catboost",
    "xgboost", "random_forest", "dummy_prior",
)
bench = {k: metrics(pairs_03b[(k, "raw")]) for k in BENCHMARK_ORDER}
fig, ax = plt.subplots(figsize=(11.7, 5.7))
ypos = np.arange(len(BENCHMARK_ORDER))
values = [bench[k]["brier"] for k in BENCHMARK_ORDER]
bar_colors = [
    COLORS["lr"] if k == "logistic_regression" else
    COLORS["cb"] if k == "catboost" else
    "#BBC6CB" if k == "dummy_prior" else "#80949D"
    for k in BENCHMARK_ORDER
]
ax.barh(ypos, values, color=bar_colors, height=.62)
ax.set_yticks(ypos, labels=[LABELS[k] for k in BENCHMARK_ORDER])
ax.invert_yaxis()
ax.set_xlim(0, max(values)*1.22)
ax.set_xlabel("Brier score · menor es mejor", labelpad=11)
ax.set_title("Siete familias, una referencia común", loc="left", pad=18)
ax.xaxis.grid(True, color=COLORS["grid"], zorder=0)
ax.set_axisbelow(True)
for y, value in zip(ypos, values, strict=True):
    ax.text(value + max(values)*.016, y, f"{value:.4f}", va="center", fontsize=10,
            color=COLORS["ink"], fontweight="semibold")
fig.subplots_adjust(left=.245, right=.96, top=.88, bottom=.17)
plt.show()
print(
    "Lectura: Brier compara calidad probabilística; "
    "el baseline evita interpretar scores sin referencia."
)
```

<div class="erl-section">04 · Calibración ERL-03C</div>
## RAW frente a SIGMOID: comparación pareada

En cada familia, RAW y SIGMOID se evalúan sobre los mismos 455 identificadores de desarrollo y las mismas cinco particiones. **Brier** y **log-loss** son métricas principales registradas previamente. El delta se define `SIGMOID − RAW`: valores negativos significan mejora; positivos, deterioro. Los métodos no se seleccionan retrospectivamente como si su comparación fuera externa.

```python
comparison = {f"{family}/{method}": metrics(pairs_03c[(family, method)])
              for family in FAMILIES_03C for method in METHODS}
rows = []
for family in FAMILIES_03C:
    for method in METHODS:
        m = comparison[f"{family}/{method}"]
        rows.append([
            LABELS[family], method.upper(), f"{m['brier']:.6f}",
            f"{m['log_loss']:.6f}", f"{100*m['sensitivity_malignant']:.2f} %",
            f"{100*m['specificity_benign']:.2f} %",
            str(m["false_negatives_malignant"]), str(m["false_positives_benign"]),
        ])
table(
    ["Familia", "Método", "Brier ↓", "Log-loss ↓", "Sensibilidad", "Especificidad", "FN", "FP"],
    rows,
    "Tabla 2. Comparación pareada RAW / SIGMOID de dos familias",
)
for family in FAMILIES_03C:
    raw, sig = comparison[f"{family}/raw"], comparison[f"{family}/sigmoid"]
    for metric, report_key in (("brier", "delta_brier"), ("log_loss", "delta_log_loss")):
        delta = sig[metric] - raw[metric]
        expected = report_03c["comparison"][f"{family}/sigmoid_minus_raw"][report_key]
        assert math.isclose(delta, expected, rel_tol=1e-10, abs_tol=1e-12)
print("Los deltas RAW/SIGMOID coinciden con el protocolo congelado.")
```

```python
def paired_effect_chart(metric: str, multiplier: float, units: str, title: str) -> None:
    arr = [multiplier * (comparison[f"{k}/sigmoid"][metric] -
                         comparison[f"{k}/raw"][metric]) for k in FAMILIES_03C]
    extent = max(abs(x) for x in arr) * 1.32
    fig, ax = plt.subplots(figsize=(11.3, 4.0))
    for index, (value, family) in enumerate(zip(arr, FAMILIES_03C, strict=True)):
        color = COLORS["lr"] if family == "logistic_regression" else COLORS["cb"]
        ax.barh(index, value, color=color, height=.43)
        direction = "MEJORA" if value < 0 else "EMPEORA" if value > 0 else "SIN CAMBIO"
        offset = extent * .024 * (1 if value >= 0 else -1)
        ax.text(value + offset, index, f"{value:+.3f} · {direction}",
                va="center", ha="left" if value >= 0 else "right",
                color=COLORS["ink"], fontsize=10, fontweight="semibold")
    ax.set_yticks(range(len(FAMILIES_03C)), [LABELS[k] for k in FAMILIES_03C])
    ax.invert_yaxis()
    ax.axvline(0, color=COLORS["ink"], lw=1.2)
    ax.set_xlim(-extent*1.22, extent*1.22)
    ax.set_xlabel(f"Cambio {units} · SIGMOID − RAW (negativo = mejora)", labelpad=9)
    ax.set_title(title, loc="left", pad=15)
    ax.xaxis.grid(True, color=COLORS["grid"])
    ax.set_axisbelow(True)
    fig.subplots_adjust(left=.24, right=.97, top=.80, bottom=.27)
    plt.show()


paired_effect_chart("brier", 1000, "Brier × 1.000", "Calibración · efecto sobre Brier")
paired_effect_chart("log_loss", 1000, "log-loss × 1.000", "Calibración · efecto sobre log-loss")
print("Interpretación: LR mejora ligeramente ambas pérdidas; CatBoost empeora ambas.")
```

<div class="erl-insight"><strong>Lectura ejecutiva.</strong> En regresión logística, SIGMOID cambia el Brier en −0,000613 y el log-loss en −0,000603, con un falso negativo menos. En CatBoost, SIGMOID empeora Brier (+0,000715) y log-loss (+0,007055), sin modificar los cinco falsos negativos. Son diferencias observadas en el mismo conjunto de desarrollo, no pruebas de superioridad clínica ni de significación estadística.</div>

<span class="erl-caption">Convención visual permanente: <strong>azul = regresión logística</strong>, <strong>coral = CatBoost</strong>; trazo continuo = RAW y discontinuo = SIGMOID. La dirección de mejora siempre se indica mediante signo y texto, nunca únicamente mediante color.</span>

<div class="erl-section">05 · Fiabilidad y heterogeneidad</div>
## Curvas de fiabilidad y comportamiento por fold

Las curvas utilizan cinco grupos de igual tamaño de los OOF externos. Son **diagnósticos descriptivos**, no una demostración de utilidad clínica ni de calibración adecuada en otra población. Los empates de score pueden cruzar los límites de los grupos.

```python
def reliability_bins(obs: list[dict]) -> tuple[np.ndarray, np.ndarray]:
    sorted_obs = sorted(obs, key=lambda r: r["p_malignant"])  # estable
    parts = np.array_split(np.arange(len(sorted_obs)), 5)
    pred = np.array([np.mean([sorted_obs[j]["p_malignant"] for j in part])
                     for part in parts])
    actual = np.array([np.mean([sorted_obs[j]["y_true_malignant"] for j in part])
                       for part in parts])
    return pred, actual


fig, ax = plt.subplots(figsize=(11.5, 6.1))
ax.plot([0, 1], [0, 1], color=COLORS["reference"], linestyle=(0, (5, 4)),
        lw=1.2, label="Calibración ideal")
styles = {
    ("logistic_regression", "raw"): (COLORS["lr"], "o", "-"),
    ("logistic_regression", "sigmoid"): (COLORS["lr"], "s", "--"),
    ("catboost", "raw"): (COLORS["cb"], "^", "-"),
    ("catboost", "sigmoid"): (COLORS["cb"], "D", "--"),
}
for key, (color, marker, linestyle) in styles.items():
    pred, observed = reliability_bins(pairs_03c[key])
    recorded = report_03c["comparison"][f"{key[0]}/{key[1]}"]["calibration"][
        "reliability_equal_count_bins"
    ]
    assert np.allclose(pred, [q["mean_predicted"] for q in recorded], atol=1e-12)
    assert np.allclose(observed, [q["mean_observed"] for q in recorded], atol=1e-12)
    ax.plot(pred, observed, color=color, marker=marker, linestyle=linestyle,
            markersize=7, lw=2.2, label=f"{LABELS[key[0]]} · {key[1].upper()}")
ax.set(xlim=(-.02, 1.02), ylim=(-.02, 1.02),
       xlabel="Probabilidad pronosticada media de malignidad",
       ylabel="Proporción observada de malignidad")
ax.set_aspect("equal", adjustable="box")
ax.set_title("Fiabilidad de la probabilidad pronosticada", loc="left", pad=18)
ax.legend(loc="center left", bbox_to_anchor=(1.02, .5), fontsize=10, frameon=False)
ax.grid(color=COLORS["grid"], alpha=.95)
fig.subplots_adjust(left=.10, right=.66, top=.88, bottom=.15)
plt.show()
print("Grupos de 91 observaciones; proximidad a la diagonal no equivale a validación externa.")
```

## Qué se aprendió y qué permanece abierto

**Lectura sustentada:**

- En el conjunto de desarrollo, **LR SIGMOID** reduce ligeramente Brier y log-loss respecto de LR RAW, con seis FN y cinco FP frente a siete FN y seis FP. No se demuestra significación estadística ni valor clínico.
- En **CatBoost**, SIGMOID aumenta Brier y log-loss; las 455 decisiones binarias son iguales a RAW con los umbrales reaprendidos.
- Las probabilidades RAW de LR y CatBoost son idénticas, observación a observación, a sus equivalentes ERL-03B. Una política distinta de umbrales explica las diferencias de decisiones LR.
- No se interpretará una curva de cinco bins como prueba de calibración ideal. Los interceptos y pendientes descriptivos no son validación externa.

**Limitaciones y amenazas a la validez:** dataset único, 455 observaciones de desarrollo, selección exploratoria de familias efectuada antes de esta comparación, incertidumbre por folds, posible optimismo por selección, registro de operaciones internas no equivalente a inspección exhaustiva de cada llamada de librería. Las advertencias `FutureWarning` de scikit-learn se mantienen como deuda de compatibilidad sin alterar retrospectivamente el experimento.

<div class="erl-danger"><strong>No inferir:</strong> seguridad clínica, desempeño prospectivo, certificación TRIPOD+AI/PROBAST+AI, utilidad para tamizaje ni superioridad general de SIGMOID.</div>
